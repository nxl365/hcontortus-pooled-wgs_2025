# To compare the three window sizes (1 kb, 10 kb, 100 kb) with corresponding step sizes (0.5 kb, 5 kb, 50 kb) 

To compare the three window-size settings, the total number of windows and the mean number of SNPs per window were calculated. For each farm population pair, the interquartile range (IQR) of FST values across all genomic windows was first calculated, and then the median of these pair-specific IQRs was used to summarise FST variability for each window size.

## assesses FST variability for each sliding-window size

1. Calculates the IQR of FST values for each farm pair.
2. Calculates the median of these pair-specific IQRs across all farm pairs to summarize overall FST variability for each window size.

```python
## function to calculate IQR of FST for each pair in each windowsize
def fst_iqr(df):
    fst_cols = df.columns[5:]
    return df[fst_cols].quantile(0.75) - df[fst_cols].quantile(0.25)

iqr_1kb = fst_iqr(data_1)
iqr_10kb = fst_iqr(data_10)
iqr_100kb = fst_iqr(data_100)

```

```python
iqr_1kb
```

```python
# median of all pairs' IQRs in each windowsize

print("Median FST IQR of all pairs in each windowsize FST data")
print("1 kb:", iqr_1kb.median().round(3))
print("10 kb:", iqr_10kb.median().round(3))
print("100 kb:", iqr_100kb.median().round(3))
```
